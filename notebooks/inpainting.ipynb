```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


def show_image(image, title="Image"):
    # Convert the image to RGB (OpenCV uses BGR)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.imshow(image)
    plt.axis("off")
    plt.title(title)
    plt.show()


# Load the image
image = cv2.imread(
    "../data/labeled/images/cocci_leuconostoc_mesenteroides_23_default_mmrsbpb_3_07102022120841.jpg"
)

show_image(image, title="Original Image")
```

```python
mask = np.zeros(image.shape[:2], np.uint8)
mask[1430:2000, 1400:1980] = 255

# Perform inpainting
inpainting_result = cv2.inpaint(image, mask, inpaintRadius=3, flags=cv2.INPAINT_TELEA)

show_image(inpainting_result, title="Inpainting Result")
```

```python
blurred_image = cv2.GaussianBlur(inpainting_result, (21, 21), 0)
show_image(blurred_image, title="Blurred Image")
```
