```python
from pathlib import Path

available_masks = sorted(list(Path("../data/labeled/masks/").glob("*.png")))

val_split = 0.25
test_split = 0.25

val_count = int(len(available_masks) * val_split)
test_count = int(len(available_masks) * test_split)

train_count = len(available_masks) - val_count - test_count

train_count, val_count, test_count
```

```python
import random

random.seed(42)
random.shuffle(available_masks)

train_masks = available_masks[:train_count]
val_masks = available_masks[train_count : train_count + val_count]
test_masks = available_masks[train_count + val_count :]

len(train_masks), len(val_masks), len(test_masks)
```

```python
import json

dict = {
    "train": [mask.stem for mask in train_masks],
    "val": [mask.stem for mask in val_masks],
    "test": [mask.stem for mask in test_masks],
}

with open("../data/labeled/splits.json", "w") as f:
    json.dump(dict, f, indent=4)
```
