```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Load the image files
img1 = mpimg.imread("coco_examples/Untitled.png")
img2 = mpimg.imread("coco_examples/Untitled2.png")
img4 = mpimg.imread("coco_examples/Untitled4.png")

# Create a new figure and set the figure size
fig = plt.figure(figsize=(7, 2))  # Adjust the figsize as needed

# Create a grid of subplots with 1 row and 4 columns
ax1 = plt.subplot(1, 3, 1)
ax2 = plt.subplot(1, 3, 2)
ax4 = plt.subplot(1, 3, 3)

# Display the images in the subplots
ax1.imshow(img1)
ax2.imshow(img2)
ax4.imshow(img4)

# Remove the x and y ticks for each subplot
ax1.axis("off")
ax2.axis("off")
ax4.axis("off")

# Display the figure
plt.tight_layout()
plt.savefig("chart_coco_examples.png", dpi=500)
plt.show()
```
