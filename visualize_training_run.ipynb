```python
import matplotlib.pyplot as plt
import seaborn as sns
import scienceplots

plt.style.use(["science", "nature"])
```

```python
import pandas as pd

# Load the CSV file to examine its contents
file_path = "training_run_20240427_2343.csv"
data = pd.read_csv(file_path)

data.head()
```

```python
# Plotting
fig, ax = plt.subplots(1, 3, figsize=(6, 2))

# Filter data for the agar_pretraining and final_training stages
agar_pretraining_data = data[data["stage"] == "agar_pretraining"]
kfold_data = (
    data[data["stage"] == "kfold"]
    .groupby("epoch")
    .mean(numeric_only=True)
    .reset_index()
)
final_training_data = data[data["stage"] == "final_training"]

# Agar pretraining loss plot
ax[0].plot(
    agar_pretraining_data["epoch"],
    agar_pretraining_data["train_loss"],
    label="Train Loss",
    color="blue",
)
ax[0].set_title("Losses (Agar Pretraining)")
ax[0].set_xlabel("Epoch")
ax[0].set_ylabel("Loss")

ax[0].plot(
    agar_pretraining_data["epoch"],
    agar_pretraining_data["val_loss"],
    label="Validation Loss",
    color="red",
)
ax[0].set_xlabel("Epoch")
ax[0].set_ylabel("Loss")
ax[0].legend()
ax[0].grid(True)

# Training loss plot
ax[1].plot(
    kfold_data["epoch"], kfold_data["train_loss"], label="Train Loss", color="blue"
)
ax[1].set_xlabel("Epoch")
ax[1].set_ylabel("Loss")
ax[1].legend()

ax[1].plot(
    kfold_data["epoch"], kfold_data["val_loss"], label="Validation Loss", color="red"
)
ax[1].set_title("K-Fold Losses")
ax[1].set_xlabel("Epoch")
ax[1].set_ylabel("Loss")
ax[1].legend()
ax[1].grid(True)

ax[1].set_ylim(0.18, 0.5)


# Final training loss plot
ax[2].plot(
    final_training_data["epoch"],
    final_training_data["train_loss"],
    label="Train Loss",
    color="green",
)
ax[2].set_title("Final Training Loss")
ax[2].set_xlabel("Epoch")
ax[2].set_ylabel("Loss")
ax[2].legend()
ax[2].grid(True)
ax[2].set_ylim(0, 0.5)

plt.tight_layout()
plt.savefig("chart_losses.png", dpi=500)
plt.show()
```

```python
# Plotting validation metrics for the kfold stage
fig, ax = plt.subplots(1, 1, figsize=(3, 2))


# Average Precision
ax.plot(
    kfold_data["epoch"],
    kfold_data["val_avg_precision_iou_50"],
    label="Avg Precision (IoU=0.50)",
    color="purple",
)
# Average Recall
ax.plot(
    kfold_data["epoch"],
    kfold_data["val_avg_recall_iou_50"],
    label="Avg Recall (IoU=0.50)",
    color="green",
)

ax.set_xlabel("Epoch")
ax.legend()
ax.grid(True)

plt.tight_layout()
plt.savefig("chart_val_metrics.png", dpi=500)
plt.show()
```
