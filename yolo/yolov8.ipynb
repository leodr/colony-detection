```python
from ultralytics import YOLO
import torch

model = YOLO("yolov8x.pt")

device = "cpu"
if torch.cuda.is_available():
    device = "cuda"
elif torch.backends.mps.is_available():
    device = "mps"
print(f"Using {device}")

model.train(data="./data.yaml", epochs=100, imgsz=1024, device=device, batch=-1)
```

```python
from PIL import Image

img_path = "../datasets/photobox_yolo/images/test/rod_lactobacillus_rennongilfy_35_default_mmrsagar_3_07102022120149.jpg"

result = model(img_path)

im_array = result[0].plot(labels=False)
im = Image.fromarray(im_array[..., ::-1])
im.save("result.jpg")
```
