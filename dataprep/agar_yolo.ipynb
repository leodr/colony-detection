```python
from ultralytics.data.converter import convert_coco
from pathlib import Path
from shutil import copy2
from tqdm.auto import tqdm

coco_path = Path("../datasets/agar_coco")
yolo_path = Path("../datasets/agar_yolo")

convert_coco(labels_dir=coco_path, save_dir=yolo_path, cls91to80=False)

yolo_label_folder = yolo_path / "labels"

label_folders = [x.name for x in yolo_label_folder.iterdir() if x.is_dir()]

for folder in label_folders:
    (yolo_path / "images" / folder).mkdir(parents=True, exist_ok=True)

for folder in label_folders:
    files = list((yolo_label_folder / folder).glob("*.txt"))

    for file in tqdm(files):
        id = file.stem

        image_path = coco_path / "data" / f"{id}.jpg"

        copy2(image_path, yolo_path / "images" / folder / f"{id}.jpg")
```
